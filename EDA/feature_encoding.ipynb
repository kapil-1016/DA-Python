## Label Encoding

Helps us to transform categorical variables into numerical variables by assigning as number values of each of the categories.

```python
from sklearn import preprocessing

le = preprocessing.LabelEncoder()

```

```python
df['Gender_labels'] = le.fit_transform(df.Gender.values)
# fit → learn all the unique categories (e.g., "Male", "Female")
# transform → convert those categories into numbers
```

```python
df.head()
```

```python
df.Gender_labels.value_counts()
```

## One Hot Encoding

Its used when independent variables are nominal.

creates K diff columns each for a category.

replace one column with 1 and rest are 0s, 0 -> absence & 1 -> presence

```python
# pd.get_dummies(), a Pandas function used for one-hot encoding categorical variables.
# it onverts categorical text/data into binary (0/1) columns.
one_hot = pd.get_dummies(df['Geography'])
```
